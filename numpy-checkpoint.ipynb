```python
#array stacking and splitting
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
print(np.vstack((a,b))) # vertical stacking : row wise stacking
print(np.hstack((a,b)))  # Horizontal stacking: column wise stacking
```

```python
c=np.array([[1,2,3],[4,5,6]]) 
print(c)
```

```python
split=np.hsplit(c,3)
for s in split:
    print(s)
split2=np.vsplit(c,2)
for s in split2:
    print(s)
```
